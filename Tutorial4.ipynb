Q2The quarter values are stored in column names instead of a single Quarter column. Sales values are spread across multiple columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Country": ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000, 10000, 14000],
    "Q2_Sales": [15000, 11000, 12500],
    "Q3_Sales": [14000, 13000, 15000],
    "Q4_Sales": [18000, 16000, 17000]
})

tidy_df = df.melt(
    id_vars="Country",
    var_name="Quarter",
    value_name="Sales"
)

# 将 Q1_Sales 等名称改为 Q1、Q2、Q3、Q4
tidy_df["Quarter"] = tidy_df["Quarter"].str.replace(
    "_Sales", "", regex=False
)

# 按原来的国家顺序排列，每个国家显示四个季度
tidy_df["Country"] = pd.Categorical(
    tidy_df["Country"],
    categories=df["Country"],
    ordered=True
)

tidy_df = tidy_df.sort_values(
    ["Country", "Quarter"]
).reset_index(drop=True)

print(tidy_df.to_string(index=False))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
}

df = pd.DataFrame(data)

# 1. 平均学习时间
average_hours = df["Study_Hours"].mean()
print(f"Average study time: {average_hours:.2f} hours")

# 2. 平均成绩
average_score = df["Score"].mean()
print(f"Average score: {average_score:.2f}")

# 3. 成绩最高的学生
highest = df.loc[df["Score"].idxmax()]
print(
    f"Highest score: Student {highest['Student']}, "
    f"Score = {highest['Score']}"
)

# 4. 学习时间最长的学生
most_hours = df.loc[df["Study_Hours"].idxmax()]
print(
    f"Most study hours: Student {most_hours['Student']}, "
    f"Hours = {most_hours['Study_Hours']}"
)

# 5. 学习时间与成绩的关系
print("Longer study hours are associated with higher scores.")

# 6. 散点图
plt.figure(figsize=(7, 5))

plt.scatter(
    df["Study_Hours"],
    df["Score"],
    color="royalblue",
    s=70
)

# 标注每个点对应的学生
for _, row in df.iterrows():
    plt.annotate(
        row["Student"],
        (row["Study_Hours"], row["Score"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
