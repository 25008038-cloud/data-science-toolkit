```python
df = pd.read_csv('/content/salary.csv')
```

```python
df =df.rename(columns={
  df.columns[5]:'Context',
  df.columns[6]:'Salary',
  df.columns[7]:'Compensation',
  df.columns[9]:'Other currency',
  df.columns[10]:'Salary context',
  df.columns[17]:'Gender'
})
```

```python
df['Salary'] = df['Salary'].replace(',','',regex=True)
df['Salary'] = df['Salary'].astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
df['Compensation'].isna().sum()
```

```python
print('--Columns and the number of missing values--')
for column_name in df.columns:
    n_missing = df[column_name].isna().sum()
    print(f'{column_name}:{n_missing}')
```

```python
df = df.drop(columns=[
    'Context',
    'Other currency',
    'Salary context'
], errors='ignore')
df.head()
```

```python
df.dropna()
df.dropna(thresh=2)
df.dropna(how='all')
print("before:",len(df))
df_dropped =df.dropna()
print("after:",len(df_dropped))
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='most_frequent')
df_filled = df.copy()
df_filled[df.columns] = imputer.fit_transform(df)
print(df_filled.isna().sum())
```

```python
from scipy import stats
df_no_outliers = df[df['Salary'] <= 1000000]
print(len(df_no_outliers))
```

```python
df['Salary'].describe()
df_valid = df[df['Salary']>0]
print(len(df_valid))
```

```python
print('--Unique Values--')
for column_name in df.columns:
    unique_values = df[column_name].unique()
    print(f'{column_name}:{unique_values}')
```
