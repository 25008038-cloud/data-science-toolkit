### 6 & 8. Missing and Null Values
Let's check if there are any missing/null values, and identify their location if present.

```python
null_counts = df.isnull().sum()
print("Null counts per column:")
print(null_counts)

# Find exact row indices for any null values
if null_counts.sum() > 0:
    null_rows = df[df.isnull().any(axis=1)]
    print("\nRows with null values:")
    display(null_rows)
else:
    print("\nNo missing/null values found in the entire dataset!")
```
